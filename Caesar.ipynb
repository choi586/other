```python
# 알파벳 하나 변환하기
char = 'C'
shift = 3
start = ord('A') # 65

new_char = chr((ord(char) - start + shift) % 26 + start)
# C(67)의 경우 = new_char = chr((67 - 65 + 3) % 26 + 65) = 70 = F
# Z(90)의 경우 = new_char = chr((90 - 65 + 3) % 26 + 65) = 67 = C

print(new_char)
```

```python
# 암호화 프로그램
text = input("암호화할 텍스트를 입력하세요: ")
shift = int(input("이동할 칸 수를 입력하세요 (정수): "))

result = ""

for char in text:
    if char.isalpha():  # 알파벳인지 확인
        if char.islower():
            start = ord('a')
        else:
            start = ord('A')
        new_char = chr((ord(char) - start + shift) % 26 + start)
        result += new_char
    else:
        result += char  # 알파벳이 아니면 그대로 둠

print(f"암호화 결과: {result}")
```

```python
# 복호화 프로그램
text = input("복호화할 텍스트를 입력하세요: ")
shift = int(input("이동할 칸 수를 입력하세요 (정수): "))

# 복호화는 shift 방향을 반대로 바꿔줌
shift = -shift

result = ""

for char in text:
    if char.isalpha():  # 알파벳인지 확인
        if char.islower():
            start = ord('a')
        else:
            start = ord('A')
        new_char = chr((ord(char) - start + shift) % 26 + start)
        result += new_char
    else:
        result += char  # 알파벳이 아니면 그대로 둠

print(f"복호화 결과: {result}")
```
